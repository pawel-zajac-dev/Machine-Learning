# Handwritten Digit Recognition using SVM

### 1. The MNIST Dataset
The MNIST dataset is a classic machine learning benchmark. It consists of thousands of grayscale images of handwritten digits (from 0 to 9). 
* Each image is 28x28 pixels, which is flattened into a 1D array of 784 features.
* In our specific CSV format, column 0 contains the target labels, and columns 1-784 contain the pixel intensity values ranging from 0 to 255.

### 2. Support Vector Machines (SVM) & Prerequisites
SVM is a powerful supervised algorithm that finds the optimal hyperplane to separate different classes by maximizing the margin between them.
* **Kernel Trick:** By default, `SVC()` uses the RBF (Radial Basis Function) kernel, which can handle non-linear relationships by mapping data into higher dimensions.
* **Why Normalization is CRUCIAL:** SVM optimization relies heavily on calculating distances between data points. If features are on different scales (e.g., one feature ranges from 0-1 and another from 0-1000), the larger feature will dominate the distance metric, ruining the model. 
* Therefore, we **must scale the data**. For image pixels (0-255), dividing by 255 is a standard and effective normalization technique (Min-Max scaling).

```python
import pandas as pd
import numpy as np
from sklearn.utils import shuffle

def getKaggleMNIST():
    # MNIST data:
    # column 0 is labels
    # column 1-784 is data, with values 0 .. 255
    # total size of CSV: (42000, 784)
    train = pd.read_csv('mnist.csv').values.astype(np.float32)
    train = shuffle(train)
    
    Xtrain = train[:-1000,1:]
    Ytrain = train[:-1000,0].astype(np.int32)
    
    Xtest = train[-1000:,1:]
    Ytest = train[-1000:,0].astype(np.int32)
    
    # scale the data (Crucial for SVM!)
    Xtrain /= 255.
    Xtest /= 255.
    
    # scaler = StandardScaler()
    # Xtrain = scaler.fit_transform(Xtrain)
    # Xtest = scaler.transform(Xtest)
    
    return Xtrain, Ytrain, Xtest, Ytest
```

### 3. Model Training and Evaluation
We will now load the preprocessed data, initialize the default Support Vector Classifier (`SVC()`), and measure how long it takes to fit the model and predict the scores. Note that SVMs can be computationally expensive (time complexity scales quadratically with the number of samples).

```python
from sklearn.svm import SVC
from datetime import datetime

# get the data: https://www.kaggle.com/c/digit-recognizer
Xtrain, Ytrain, Xtest, Ytest = getKaggleMNIST()

model = SVC()

t0 = datetime.now()
model.fit(Xtrain, Ytrain)
print("train duration:", datetime.now() - t0)

t0 = datetime.now()
print("train score:", model.score(Xtrain, Ytrain), "duration:", datetime.now() - t0)

t0 = datetime.now()
print("test score:", model.score(Xtest, Ytest), "duration:", datetime.now() - t0)
```
